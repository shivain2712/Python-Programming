```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
plt.plot(df.job_posted_date, df.job_posted_date)
```

```python
date_counts = df.job_posted_date.value_counts()
date_counts = date_counts.sort_index()
#type(date_counts)
```

```python
plt.plot(date_counts.index, date_counts)
#not optimal
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.month
```

```python
month_counts = df.job_posted_month.value_counts().sort_index()
```

```python
plt.plot(month_counts.index, month_counts.values)
#better
```

```python
job_counts_1 = job_counts.sort_values(ascending=True)
job_counts_1
```

```python
plt.bar(job_counts_1.index, job_counts_1.values, label='Job Counts')
plt.ylabel('COUNT')
plt.xlabel('JOB TITLE')
plt.xticks(rotation=45, ha='right') #label formatting for x axis
plt.legend()
plt.title('Job Details')
```
