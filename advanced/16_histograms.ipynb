```python
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset
import ast 
from adjustText import adjust_text 
import seaborn as sns

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_india_anal = df[(df['job_country'] == 'India') & (df['job_title_short'] == "Data Analyst")].copy() 
```

```python
df_india_anal = df_india_anal.dropna(subset='salary_year_avg')
```

```python
df_india_anal['salary_year_avg'].plot(kind='hist', bins=25, edgecolor='black')
ax = plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos : f'${int(x/1000)}K'))
plt.title('Distribution of Salaries for Data Analysts in India')
plt.show()
```

```python
sns.set_theme(style='ticks')
sns.displot(df_india_anal['salary_year_avg'], kind='kde', fill=True)
#df_india_anal['salary_year_avg'].plot(kind='hist', bins=25, edgecolor='black')
ax = plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos : f'${int(x/1000)}K'))
plt.title('Distribution of Salaries for Data Analysts in India')
plt.ylabel('Number of Jobs')
plt.xlabel('Yearly Salary')
plt.show()
```
