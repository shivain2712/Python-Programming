```python
#import libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#load dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
median_salary_year = df['salary_year_avg'].median()
median_salary_hour = df['salary_hour_avg'].median()
```

```python
df_filled = df #create a copy of the og df

#fill in median values in place of NaN
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df_filled.loc[:, 'salary_year_avg':'salary_hour_avg'] #check
```

```python
df_unique = df_filled

df_unique = df_unique.drop_duplicates()
```

```python
print('length of original df is : ', len(df) )
print('length of df_unique is : ', len(df_unique) )
print('rows dropped : ', len(df)-len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name']) #clean up and drop repetitive job postings from the same company
 
print('length of original df is : ', len(df) )
print('length of df_unique is : ', len(df_unique) )
print('rows dropped : ', len(df)-len(df_unique))
```
